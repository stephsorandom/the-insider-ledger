Downloading ZIP files with Python

```python
import requests
```

```python
url = 'https://www.sec.gov/files/structureddata/data/insider-transactions-data-sets/2026q1_form345.zip'

# SEC requires a User-Agent identifying the requester or it returns 403 Forbidden
headers = {
    'User-Agent': 'Stephanie Mitchell stephmitch023@gmail.com'
}

# Pull the quarterly ZIP; response holds status + raw file content
response = requests.get(url, headers=headers)
```

```python
# Validate get request ran properly
print(response.status_code)
```

```python
# Save the downloaded bytes to disk as a real ZIP file, so we can unzip it next
with open('2026q1_form345.zip', 'wb') as f:
    f.write(response.content)
```

```python
import zipfile
```

```python
# Extract into a quarter-specific subfolder, since all 8 filenames repeat every quarter
with zipfile.ZipFile('2026q1_form345.zip', 'r') as z:
    z.extractall('raw_data/2026q1')
```

```python
# List of dictionaries to loop through and download all quarters
quarters = [
    {'year': 2026, 'q': 2, 'path': 'datastandardsinnovation'},
    {'year': 2026, 'q': 1, 'path': 'structureddata'},
    {'year': 2025, 'q': 4, 'path': 'structureddata'},
    {'year': 2025, 'q': 3, 'path': 'structureddata'},
    {'year': 2025, 'q': 2, 'path': 'structureddata'},
    {'year': 2025, 'q': 1, 'path': 'structureddata'},
    {'year': 2024, 'q': 4, 'path': 'structureddata'},
    {'year': 2024, 'q': 3, 'path': 'structureddata'},
    {'year': 2024, 'q': 2, 'path': 'structureddata'},
    {'year': 2024, 'q': 1, 'path': 'structureddata'},
    {'year': 2023, 'q': 4, 'path': 'structureddata'},
    {'year': 2023, 'q': 3, 'path': 'structureddata'}    
]
```

```python
# Loop through every quarter in our list and repeat the download + unzip steps
for entry in quarters:
    # Pull the 3 pieces of info out of this quarter's dictionary
    year = entry['year']
    q = entry['q']
    path = entry['path']

    # Build this quarter's URL using an f-string
    url = f'https://www.sec.gov/files/{path}/data/insider-transactions-data-sets/{year}q{q}_form345.zip'

    # Download it (headers dictionary already exists from before)
    response = requests.get(url, headers=headers)

    # Save it to a filename that changes each loop
    filename = f'{year}q{q}_form345.zip'
    with open(filename, 'wb') as f:
        f.write(response.content)

    # Unzip into a quarter-specific subfolder
    with zipfile.ZipFile(filename, 'r') as z:
        z.extractall(f'raw_data/{year}q{q}')

    # Confirm each quarter succeeded before moving to the next
    print(f'{year}Q{q}: status {response.status_code}')
```
