## Cleaning: RPTOWNER_RELATIONSHIP (multi-role field)

SEC stores multiple insider roles as one comma-separated string 
(e.g. "Director,Officer").<br> 
Splitting into separate boolean flag 
columns so each role is independently queryable, without losing 
any role a person actually holds.

```python
reportingowner['RPTOWNER_RELATIONSHIP'].unique()
```

```python
# Verifying NAN for cleaning design
reportingowner['RPTOWNER_RELATIONSHIP'].isna().sum()
```

```python
# Flag each role from RPTOWNER_RELATIONSHIP. Two steps per flag:
# 1) .str.contains() checks if the role name appears anywhere in the (possibly combined) text
# 2) .astype('boolean') converts to a nullable boolean type so missing relationships can stay NA instead of being forced to False
# Then .loc explicitly sets the True NA rows (no relationship listed at all) to NA, since step 1 alone doesn't reliably preserve that

reportingowner['is_officer'] = reportingowner['RPTOWNER_RELATIONSHIP'].str.contains('Officer').astype('boolean')
reportingowner.loc[reportingowner['RPTOWNER_RELATIONSHIP'].isna(), 'is_officer'] = pd.NA

reportingowner['is_director'] = reportingowner['RPTOWNER_RELATIONSHIP'].str.contains('Director').astype('boolean')
reportingowner.loc[reportingowner['RPTOWNER_RELATIONSHIP'].isna(), 'is_director'] = pd.NA

reportingowner['is_ten_percent_owner'] = reportingowner['RPTOWNER_RELATIONSHIP'].str.contains('TenPercentOwner').astype('boolean')
reportingowner.loc[reportingowner['RPTOWNER_RELATIONSHIP'].isna(), 'is_ten_percent_owner'] = pd.NA

reportingowner['is_other'] = reportingowner['RPTOWNER_RELATIONSHIP'].str.contains('Other').astype('boolean')
reportingowner.loc[reportingowner['RPTOWNER_RELATIONSHIP'].isna(), 'is_other'] = pd.NA
```

### Data Quality Note: Missing `TRANS_PRICEPERSHARE`

Of the 28,757 filtered transaction records, 58 (~0.2%) have a missing `TRANS_PRICEPERSHARE` value.

The missing values were investigated by `TRANS_CODE` and were not isolated to a single transaction type:

* **42 Purchases (`P`)**
* **16 Sales (`S`)**

A review of the available ownership fields showed that some missing-price purchase records were associated with indirect ownership structures, including family trusts and IRA rollovers, based on the `NATURE_OF_OWNERSHIP` field. The missing-price sales did not show a clear, consistent pattern in the available columns.

A definitive explanation for the missing prices would likely require reviewing the transaction footnotes. However, the `FOOTNOTES` file was intentionally outside the scope of this project because it was not required to answer the five defined business questions.

**Decision:** The 58 missing prices were left as `NaN` rather than imputed. There was no reliable basis in the available data to determine or estimate the actual transaction price without introducing unsupported assumptions.


## Cleaning function: process one quarter at a time

Wraps the full cleaning pipeline into one reusable function, `clean_quarter(quarter)`. <br>
The function reads the three files for a given quarter, standardizes boolean fields,<br> 
splits `RPTOWNER_RELATIONSHIP` into role flags, drops `_FN` columns, and filters transactions to purchases and sales (`P`/`S`).

The function is then called once for each quarter, avoiding the need to repeat the same cleaning logic 12 times.



```python
def clean_quarter(quarter):
    # quarter is a placeholder — a string like "2026q1", handed in each time this function is called

    # Read the 3 files for this quarter (reuse your existing read logic, 
    # but swap the hardcoded "2026q1" for the quarter parameter instead)
    submission = pd.read_csv(f"raw_data/{quarter}/SUBMISSION.tsv", sep='\t', dtype={'ISSUERCIK': str})
    reportingowner = pd.read_csv(f'raw_data/{quarter}/REPORTINGOWNER.tsv', sep='\t', dtype={'RPTOWNERCIK': str})
    nonderiv_trans = pd.read_csv(f'raw_data/{quarter}/NONDERIV_TRANS.tsv', sep='\t')

    # ... apply your existing cleaning steps here (booleans, role flags, _FN drop, P/S filter) ...
    submission['NOT_SUBJECT_SEC16'] = submission['NOT_SUBJECT_SEC16'].replace({
        'false' : False,
        0 : False,
        '0' : False,
        'true' : True,
        1 : True,
        '1' : True
    })

    submission['AFF10B5ONE'] = submission['AFF10B5ONE'].replace({
        'false' : False,
        '0' : False,
        'true' : True,
        '1' : True
    })

    nonderiv_trans['EQUITY_SWAP_INVOLVED'] = nonderiv_trans['EQUITY_SWAP_INVOLVED'].replace({
        'false' : False,
        '0': False,
        '1' : True
    })

    nonderiv_trans = nonderiv_trans[nonderiv_trans['TRANS_CODE'].isin(['P', 'S'])]


    reportingowner['is_officer'] = reportingowner['RPTOWNER_RELATIONSHIP'].str.contains('Officer').astype('boolean')
    reportingowner.loc[reportingowner['RPTOWNER_RELATIONSHIP'].isna(), 'is_officer'] = pd.NA

    reportingowner['is_director'] = reportingowner['RPTOWNER_RELATIONSHIP'].str.contains('Director').astype('boolean')
    reportingowner.loc[reportingowner['RPTOWNER_RELATIONSHIP'].isna(), 'is_director'] = pd.NA

    reportingowner['is_ten_percent_owner'] = reportingowner['RPTOWNER_RELATIONSHIP'].str.contains('TenPercentOwner').astype('boolean')
    reportingowner.loc[reportingowner['RPTOWNER_RELATIONSHIP'].isna(), 'is_ten_percent_owner'] = pd.NA

    reportingowner['is_other'] = reportingowner['RPTOWNER_RELATIONSHIP'].str.contains('Other').astype('boolean')
    reportingowner.loc[reportingowner['RPTOWNER_RELATIONSHIP'].isna(), 'is_other'] = pd.NA

    fn_columns = nonderiv_trans.columns[nonderiv_trans.columns.str.endswith('_FN')]
    nonderiv_trans = nonderiv_trans.drop(columns=fn_columns)

    fn_columns = submission.columns[submission.columns.str.endswith('_FN')]
    submission = submission.drop(columns=fn_columns)

    fn_columns = reportingowner.columns[reportingowner.columns.str.endswith('_FN')]
    reportingowner = reportingowner.drop(columns=fn_columns)

    # Hand back all 3 cleaned tables so whoever calls this function can use them
    return submission, reportingowner, nonderiv_trans
```

```python
# Test the function on one quarter before looping across all 12
test_submission, test_reportingowner, test_nonderiv_trans = clean_quarter('2026q1')

print(test_submission.shape)
print(test_reportingowner.shape)
print(test_nonderiv_trans.shape)
```
