```python
import pandas as pd
```

```python
# Read SUBMISSION file into a DataFrame
submission = pd.read_csv(
    'raw_data/2026q1/SUBMISSION.tsv',
    sep='\t', # separator '\t' tells pandas the columns are separated by tabs, not commas (confirmed in Phase 1)
    dtype={'ISSUERCIK': str} # dtype forces specified column to stay as text (str), protecting leading zeros that a number would silently drop
)

# Quick sanity check: how many rows, and what do the first 5 look like?
print(submission.shape)
submission.head()
```

```python
# Read REPORTINGOWNER.tsv into a DataFrame.
# sep='\t' tells pandas the columns are tab-separated, not comma-separated.
# dtype forces RPTOWNERCIK to stay text, protecting leading zeros in this CIK field.
reportingowner = pd.read_csv(
    'raw_data/2026q1/REPORTINGOWNER.tsv',
    sep='\t',
    dtype={'RPTOWNERCIK': str}
)

print(reportingowner.shape)
reportingowner.head()
```

```python
# Read NONDERIV_TRANS.tsv into a DataFrame — same pattern as your last 2 reads
nonderiv_trans = pd.read_csv(
    'raw_data/2026q1/NONDERIV_TRANS.tsv',
    sep='\t'
)

print(nonderiv_trans.shape)
nonderiv_trans.head()
```

```python
# Filter nonderiv_trans down to only Purchase (P) and Sale (S) transactions.

# .isin() checks TRANS_CODE against the list (P&S) and returns True/False per row.
# Outter nonderiv_trans [] is to keep any rows that return True (P & S) and anything else is dropped
nonderiv_trans = nonderiv_trans[nonderiv_trans['TRANS_CODE'].isin(['P', 'S'])]

print(nonderiv_trans.shape)
nonderiv_trans.head()
```

```python
# Standarize NOT_SUBJECT_SEC16 in the submission table to real Boolean values
submission['NOT_SUBJECT_SEC16'] = submission['NOT_SUBJECT_SEC16'].replace({
    'false' : False,
    0 : False,
    '0' : False,
    'true' : True,
    1 : True,
    '1' : True

})

submission['NOT_SUBJECT_SEC16'].unique()
```
